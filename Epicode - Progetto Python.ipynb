```python
import pandas as pd
```

```python
covid_df = pd.read_csv('owid-covid-data.csv')
```

Si richiede di verificare le dimensioni del dataset e le diciture presenti nell'intestazione

```python
dimensioni = covid_df.shape
elementi_totali = covid_df.size
column_names = [covid_df.columns]
```

Si chiede poi per ogni continente di trovare il numero di casi totali avvenuti in quello stesso continente;
si chiede di non considerare eventuali locazioni che nel dataset non appartengono ad alcun continente

```python
case_total_by_continent = covid_df.groupby('continent')['new_cases'].sum()
```

Sempre riguardo i casi di COVID totali, si chiede di sviluppare una funzione che prenda in input il dataset e due nomi di continenti, e che ne confronti i seguenti relativi descrittori statistici: valori minimo e massimo, media, e percentuale rispetto al numero dei casi totali nel mondo (calcolati anche sulle locazioni senza indicazione di continente)

```python
data_piu_recente = covid_df['date'].max()
case_total_at_data_piu_recente = covid_df[covid_df['date'] == data_piu_recente]
case_total_worldwide = case_total_at_data_piu_recente['total_cases'].sum()

def confronto_statistiche(covid_df, continente_1, continente_2):  
    covid_df_continente_1 = covid_df[covid_df["continent"] == continente_1]
    covid_df_continente_2 = covid_df[covid_df["continent"] == continente_2] 
    
    total_cases_continente_1 = covid_df_continente_1['new_cases'].sum()
    total_cases_continente_2 = covid_df_continente_2['new_cases'].sum()
    
    desc_continente_1 = covid_df_continente_1['new_cases'].describe()
    desc_continente_2 = covid_df_continente_2['new_cases'].describe()
    
    percent_continente_1 = (total_cases_continente_1 / case_total_worldwide) * 100
    percent_continente_2 = (total_cases_continente_2 / case_total_worldwide) * 100
    
    calculations_results = pd.DataFrame({
        'Min': [desc_continente_1['min'], desc_continente_2['min']],
        'Mean': [desc_continente_1['mean'], desc_continente_2['mean']],
        'Max': [total_cases_continente_1, total_cases_continente_2],
        'Percent over worldwide': [percent_continente_1, percent_continente_2]
    }, index=[f'{continente_1} Values', f'{continente_2} Values'])
    
    return calculations_results

calculations_results = confronto_statistiche(covid_df, 'Oceania', 'Europe')
print(calculations_results)
```

Si chiede poi di effettuare lo stesso tipo di analisi – anche in questo caso sviluppando una funzione ad hoc – per il numero di vaccinazioni totali per ogni continente

```python
data_piu_recente = covid_df['date'].max()
vaccination_total_at_data_piu_recente = covid_df[covid_df['date'] == data_piu_recente]
vaccination_total_worldwide = vaccination_total_at_data_piu_recente['total_vaccinations'].sum()

def confronto_statistiche_vaccini(covid_df, continente_1, continente_2):  
    covid_df_continente_1 = covid_df[covid_df["continent"] == continente_1]
    covid_df_continente_2 = covid_df[covid_df["continent"] == continente_2] 
    
    total_vaccinations_continente_1 = covid_df_continente_1['new_vaccinations'].sum()
    total_vaccinations_continente_2 = covid_df_continente_2['new_vaccinations'].sum()
    
    max_continente_1 = total_vaccinations_continente_1.max()
    max_continente_2 = total_vaccinations_continente_2.max()
    
    min_mean_continente_1 = covid_df_continente_1['new_vaccinations'].describe()[['min', 'mean']]
    min_mean_continente_2 = covid_df_continente_2['new_vaccinations'].describe()[['min', 'mean']]
    
    percent_continente_1 = (total_vaccinations_continente_1 / vaccination_total_worldwide) * 100
    percent_continente_2 = (total_vaccinations_continente_2 / vaccination_total_worldwide) * 100
    
    calculations_results = pd.DataFrame({
        'Max Vaccinations': [max_continente_1, max_continente_2],
        'Min Vaccinations': [min_mean_continente_1['min'], min_mean_continente_2['min']],
        'Mean Vaccinations': [min_mean_continente_1['mean'], min_mean_continente_2['mean']],
        'Percent over worldwide Vaccinations': [percent_continente_1, percent_continente_2]
    }, index=[f'{continente_1} Values', f'{continente_2} Values'])
    
    return calculations_results

calculations_results_vaccini = confronto_statistiche_vaccini(covid_df, 'Oceania', 'Europe')
print(calculations_results_vaccini)
```

Sulla base dei dati ottenuti possiamo notare che le vaccinazioni medie sono pressocchè uguali tra Oceania ed Europa, mentre risulta molto più alta quella del Sud America. L'Europa, seguita dal Sud America è stato il continente col più alto numero di vaccinazioni e questa classifica rispecchia anche il numero di abitanti rispettivo.
Vedendo invece i dati sui casi covid, notiamo che l'Europa ha avuto molti più casi degli altri due paesi, avendo una media simile a quella del Sud America, nonostante quest'ultima abbia registrato un numero max molto inferiore.
